```python
import tensorflow as tf

# Load the model from .keras file
model = tf.keras.models.load_model("fer_best_model.keras")

# Verify that the model is loaded correctly
model.summary()
```

```python
# Save the model in TensorFlow SavedModel format
saved_model_path = "fer_best_model"
model.save(saved_model_path, save_format="tf")

print(f"Model successfully saved to {saved_model_path}")
```

```python
import tarfile

tar_filename = "fer_best_model.tar.gz"

# Create a tar.gz archive
with tarfile.open(tar_filename, "w:gz") as tar:
    tar.add(saved_model_path, arcname="fer_best_model")

print(f"SavedModel successfully compressed into {tar_filename}")
```

```python
with tarfile.open(tar_filename, "r:gz") as tar:
    tar.list()
```
