```python
import os
from PIL import Image

def verify_image_sizes(directory):
    for root, _, files in os.walk(directory):
        for file in files:
            img_path = os.path.join(root, file)
            img = Image.open(img_path)
            if img.size != (48, 48):
                print(f"Image {file} is {img.size}")

# Run this on your train_dir and test_dir
verify_image_sizes(train_dir)
verify_image_sizes(test_dir)
```
